## Tokenizer : How the GPT Token Sytem works

```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4")

tokens = encoding.encode("Hello, world!")
```

```python
print(tokens)
```

```python
for token_id in tokens:
    decoded_token = encoding.decode([token_id])
    print(f"{token_id} ---> '{decoded_token}'"  )
```

## Memory Leaking

```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

api_key = os.getenv("GROQ_API_KEY")
```

```python
from openai import OpenAI

groq_client = OpenAI(api_key=api_key, base_url="https://api.groq.ai/v1")
```

```python
message = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is the capital of France?"}
]
```

```python
openai_response = groq_client.chat.completions.create(
    model="gpt-4o-mini",
    messages=message,
)
print(openai_response.choices[0].message.content)
```
